# AISAM model config generator

Use this notebook to create a `model_config.json` file for training a forecaster from existing simulation data.

The generated config can be passed to:

```python
from aisam.utils.forecaster_training import train_forecaster_from_simulation

train_forecaster_from_simulation(
    simulation_paths="/path/to/simulation.pkl",
    model="/path/to/model_config.json"
)
```

````python
import json
from pathlib import Path
````

## Output location

Set the folder where `model_config.json` should be saved. This can be the same folder that contains your `simulation.pkl`, or any other experiment folder.

````python
output_folder = Path("/Users/hossein/Documents/Projects/AISAM/assets/260520")
output_folder.mkdir(parents=True, exist_ok=True)
model_config_path = output_folder / "model_config.json"
model_config_path
````

## Model type

The currently implemented standard forecaster training path supports `regressor`. Transformer/LSTM entries can be saved in the config for future support, but the current training function will raise `NotImplementedError` for them.

````python
model_type = "regressor"

# Common future options once connected to training backends:
# model_type = "transformer"
# model_type = "lstm"
````

## Training data source

These paths are saved as metadata for reproducibility. The training function still receives `simulation_paths` directly, so you can override these at call time.

````python
simulation_paths = [
    "/path/to/training_data/ccasr_YYYY-MM-DD_HH-MM-SS/simulation.pkl",
]

output_root = None  # None saves under the simulation run's models/ folder.
````

## Forecaster hyperparameters

````python
hyperparameters = {
    "past_feature_window": 20,
    "future_window": 1,
    "past_input_window": None,
    "future_input_window": None,
    "stride": 1,
    "validation_fraction": 0.2,
    "feature_species": None,
    "output_species": "F",
    "normalize": True,
    "random_state": 0,
}
````

## Evaluation and plotting options

````python
visualization = True

# For random-stim-only training, repetitive/periodic cells can still be included in evaluation.
train_random_stims_only = False
include_periodic_stims_in_validation = True
````

## Build and preview model config

````python
model_config = {
    "type": model_type,
    "hyperparameters": hyperparameters,
    "visualization": visualization,
    "train_random_stims_only": train_random_stims_only,
    "include_periodic_stims_in_validation": include_periodic_stims_in_validation,
    "data": {
        "simulation_paths": simulation_paths,
    },
}

if output_root is not None:
    model_config["output_root"] = str(output_root)

print(json.dumps(model_config, indent=2))
````

## Save model_config.json

````python
with open(model_config_path, "w") as f:
    json.dump(model_config, f, indent=2)

model_config_path
````
