```python
import numpy as np
import matplotlib.pyplot as plt
import ast
#import torch
import re
import dill as pickle 


import json
import requests
```

```python
API_URL = "http://localhost:8000/v1/chat/completions"
MODEL_NAME = "Qwen/Qwen3-32B"

def chat_with_llm(message, history):
  headers = {"Content-Type": "application/json"}
  messages = [{"role": "system", "content": "You are in control of a dynamical system. The user gives you the state of the system at every turn and you should steer it towards the user-given goal by giving a series of inputs that best suits the system."}]
  if history:
    for user_msg, bot_reply in history:
      messages.append({"role": "user", "content": user_msg})
      messages.append({"role": "assistant", "content": bot_reply})

  # Append the latest user message
  messages.append({"role": "user", "content": message})

  payload = {
  "model": MODEL_NAME,
  "messages": messages,
  "temperature": 0.2,
  "top_p": 0.9
  #'max_tokens':256,
  }

  response = requests.post(API_URL, headers=headers, data=json.dumps(payload))
  try:
    reply = response.json()["choices"][0]["message"]["content"]
  except:
    reply = "Error: Could not get response from model."

  full_reply = (messages),(reply)

  return full_reply
```

```python
ans = chat_with_llm('Explain fully how you would solve the problem of an inverted pendulum.',None)
print(ans[1])
```
